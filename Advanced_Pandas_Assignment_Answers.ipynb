**Q2 Sales Filtering Analysis**

Filter the dataset to display:

1. Orders where Sales > 3500

2. Orders where Region = "East" and Sales > 2500

Explain what insights you observe.

```python
#A2.
# Group by customer and sum the sales, then sort descending
top_customers = df.groupby('Customer')['Sales'].sum().sort_values(ascending=False)

print("--- Top 5 Customers by Total Sales ---")
display(top_customers.head(5))
```

Overall, it cuts through the noise of everyday small orders to highlight what's really driving your profits.

**Q3 Customer Purchase Analysis**

Identify the top 5 customers based on total sales.

Hint: Use GroupBy with aggregation.

```python
#A3.
# Group by customer and sum the sales, then sort descending
top_customers = df.groupby('Customer')['Sales'].sum().sort_values(ascending=False)

print("--- Top 5 Customers by Total Sales ---")
display(top_customers.head(5))
```

**Q4 Regional Sales Performance**

Calculate the total sales generated by each region.

Then identify:

* Which region generated the highest revenue?

```python
#A4.
# Calculate total sales by region
regional_sales = df.groupby('Region')['Sales'].sum()
print("--- Total Sales by Region ---")
print(regional_sales)

# Identify the region with the highest revenue
top_region = regional_sales.idxmax()
highest_revenue = regional_sales.max()
print(f"\nThe region with the highest revenue is {top_region} with total sales of {highest_revenue}.")
```

**Q5 Product Performance Analysis**

* Calculate the total quantity sold for each product.

* Which product appears to be the most popular?

```python
#A5.
# Calculate total quantity sold for each product
product_qty = df.groupby('Product')['Quantity'].sum()
print("--- Total Quantity Sold per Product ---")
print(product_qty)

# Identify the most popular product
popular_product = product_qty.idxmax()
max_qty = product_qty.max()
print(f"\nThe most popular product is {popular_product} with {max_qty} units sold.")
```

**Q6 Multi-Level Grouping**

* Group the dataset by: Region and Category

* Calculate the total sales for each combination.

* What insights can you derive from this analysis?

```python
#A6.
# Group by Region and Category, calculate total sales
multi_level_group = df.groupby(['Region', 'Category'])['Sales'].sum().reset_index()

print("--- Total Sales by Region and Category ---")
display(multi_level_group)
```

**Q7 Pivot Table Creation**

Create a pivot table showing:

* Rows → Region

* Columns → Product

* Values → Total Sales

Which region sells the most laptops?

```python
#A7.
# Create the pivot table
pivot_sales = pd.pivot_table(df, values='Sales', index='Region', columns='Product', aggfunc='sum')

print("--- Pivot Table: Region vs Product Sales ---")
display(pivot_sales)

# Find which region sells the most Laptops
top_laptop_region = pivot_sales['Laptop'].idxmax()
print(f"\nThe region that sells the most laptops is: {top_laptop_region}")
```

**Q8 Profitability Analysis**

* Calculate the average profit for each product category.

* Which category is most profitable?

```python
#A8.
# Calculate average profit for each category
avg_profit_category = df.groupby('Category')['Profit'].mean()

print("--- Average Profit by Category ---")
print(avg_profit_category)

# Identify the most profitable category
most_profitable = avg_profit_category.idxmax()
print(f"\nThe most profitable category is: {most_profitable}")
```

**Q9 Discount Impact Study**

Analyze the relationship between discount and profit.

Tasks:

* Calculate the average profit for each discount level

* Identify whether higher discounts lead to lower profit

```python
#A9.
# Calculate average profit for each discount level
discount_impact = df.groupby('Discount')['Profit'].mean()

print("--- Average Profit per Discount Level ---")
print(discount_impact)

# To formally check if higher discounts lead to lower profits, we can check the correlation
correlation = df['Discount'].corr(df['Profit'])
print(f"\nCorrelation between Discount and Profit: {correlation:.2f}")
if correlation < 0:
    print("Insight: Yes, higher discounts generally lead to lower profitability (negative correlation).")
else:
    print("Insight: No, higher discounts do not lead to lower profitability based on this dataset.")
```

**Q10 End-to-End Data Wrangling Task**

Perform the following workflow:

1. Filter the dataset where Sales > 3000

2. Group data by Region and Product

3. Calculate total sales

4. Reshape the result into a pivot table

Explain the business insights obtained from this analysis.

```python
#A10.
# 1. Filter the dataset where Sales > 3000
high_value_df = df[df['Sales'] > 3000]

# 2 & 3. Group data by Region and Product and calculate total sales
grouped_high_value = high_value_df.groupby(['Region', 'Product'])['Sales'].sum().reset_index()

# 4. Reshape the result into a pivot table
final_pivot = pd.pivot_table(grouped_high_value, values='Sales', index='Region', columns='Product', aggfunc='sum')

print("--- End-to-End Wrangling: High Value Orders Pivot Table ---")
display(final_pivot.fillna(0)) # Using fillna(0) to replace NaN with 0 for cleaner viewing
```

**Business Insight:** By filtering for only high-value orders (>$3000) and viewing them across regions and products, businesses can identify their "premium" markets. If a specific region is exclusively buying high-value Electronics, the business can focus its VIP customer retention programs and premium product advertising in that specific region.
